# Plan to preprocess based on mode later

```python
df_mode = df.groupby('state')['mode'].unique().reset_index()
df_mode
```

```python
df_mode['mode'].astype(str).unique()
```

```python
(df_mode
    .assign(mode_str=lambda x: x['mode'].astype(str))
    .groupby('mode_str')['state']
    .apply(list)
)
```

- NEW MEXICO = ['ABSENTEE' 'EARLY VOTING' 'ELECTION DAY']
- TEXAS = [nan 'EARLY VOTING' 'TOTAL VOTES']

```python
# TOTAL VOTES only
(df_mode
    .assign(mode_str=lambda x: x['mode'].astype(str))
    .groupby('mode_str')['state']
    .apply(list)
    .loc["['TOTAL VOTES']"]
)
```

```python
# Including TOTAL VOTES
df.query("mode == 'TOTAL VOTES'")['state'].unique()
```

```python
# TOTAL VOTES + other modes
set(df.query("mode == 'TOTAL VOTES'")['state'].unique()) - set((df_mode
    .assign(mode_str=lambda x: x['mode'].astype(str))
    .groupby('mode_str')['state']
    .apply(list)
    .loc["['TOTAL VOTES']"]
))
```

```python
# nan only
(df_mode
    .assign(mode_str=lambda x: x['mode'].astype(str))
    .groupby('mode_str')['state']
    .apply(list)
    .loc["[nan]"]
)
```

```python
# TOTAL VOTES only + (TOTAL VOTES + other modes) + nan only = 50
len((df_mode
    .assign(mode_str=lambda x: x['mode'].astype(str))
    .groupby('mode_str')['state']
    .apply(list)
    .loc["['TOTAL VOTES']"]
)) + len(set(df.query("mode == 'TOTAL VOTES'")['state'].unique()) - set((df_mode
    .assign(mode_str=lambda x: x['mode'].astype(str))
    .groupby('mode_str')['state']
    .apply(list)
    .loc["['TOTAL VOTES']"]
))) + len((df_mode
    .assign(mode_str=lambda x: x['mode'].astype(str))
    .groupby('mode_str')['state']
    .apply(list)
    .loc["[nan]"]
))
```

```python
# NEW MEXICO (other modes, no TOTAL VOTES)
set(df_mode['state'].unique()) - set(
    (df_mode
    .assign(mode_str=lambda x: x['mode'].astype(str))
    .groupby('mode_str')['state']
    .apply(list)
    .loc["['TOTAL VOTES']"]
)
) - (set(df.query("mode == 'TOTAL VOTES'")['state'].unique()) - set((df_mode
    .assign(mode_str=lambda x: x['mode'].astype(str))
    .groupby('mode_str')['state']
    .apply(list)
    .loc["['TOTAL VOTES']"]
))) - set(
    (df_mode
    .assign(mode_str=lambda x: x['mode'].astype(str))
    .groupby('mode_str')['state']
    .apply(list)
    .loc["[nan]"]
)
)
```
