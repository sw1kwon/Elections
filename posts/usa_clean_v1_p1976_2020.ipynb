# v1

## writein NaN

```python
df['writein'] = df['writein'].astype(str)
```

```python
df['writein'].value_counts(dropna=False)
```

```python
df.query("year == 2020 and state_po == 'MD'  and writein == 'nan'")
```

- https://en.wikipedia.org/wiki/2020_United_States_presidential_election_in_Maryland#Results
- Write-in: 20,422

```python
df.query("year == 2020 and state_po == 'MD'  and writein == 'True'")['candidatevotes'].sum()
```

```python
df['writein'] = df['writein'].mask(
    (df['year'] == 2020) &
    (df['state_po'] == "MD") &
    (df['writein'] == "nan"),
    "True"
)
```

```python
df.query("year == 2020 and state_po == 'MD'  and writein == 'True'")['candidatevotes'].sum()
```

## writein True False

- https://electionresults.dcboe.org/election_statistics/2020-General-Election
- Write-in: 3,137

```python
df.loc[(df['year'] == 2020) & (df['state_po'] == "DC")]
```

```python
df['writein'] = df['writein'].mask(
    (df['year'] == 2020) &
    (df['state_po'] == "DC") &
    (df['candidatevotes'] != 3137) &
    (df['writein'] == "True"),
    "False"
)
```
