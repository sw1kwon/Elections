# Data (President & Governor, v4.3) & Merge

```python
import pandas as pd

# 대선 데이터 14~21회 불러오기 및 연결
president_dfs = []

for i in range(14, 22):  # 14, 15, 16, 17, 18, 19, 20, 21
    url = f"https://raw.githubusercontent.com/sw1kwon/korean-elections/refs/heads/main/temp/v4_3_p/temp4_3_president_{i}.csv"
    df = pd.read_csv(url)
    president_dfs.append(df)
    print(f"대선 {i}회 데이터 로드 완료: {df.shape}")

# 대선 데이터 연결
president_combined = pd.concat(president_dfs, ignore_index=True)
print(f"대선 데이터 연결 완료: {president_combined.shape}")

# 지선 데이터 3~8회 불러오기 및 연결
governor_dfs = []

for i in range(3, 9):  # 3, 4, 5, 6, 7, 8
    url = f"https://raw.githubusercontent.com/sw1kwon/korean-elections/refs/heads/main/temp/v4_3_g/temp4_3_governor_{i}.csv"
    df = pd.read_csv(url)
    governor_dfs.append(df)
    print(f"지선 {i}회 데이터 로드 완료: {df.shape}")

# 지선 데이터 연결
governor_combined = pd.concat(governor_dfs, ignore_index=True)
print(f"지선 데이터 연결 완료: {governor_combined.shape}")

# 대선 + 지선 데이터 전체 연결
all_combined = pd.concat([president_combined, governor_combined], ignore_index=True)
print(f"전체 데이터 연결 완료: {all_combined.shape}")

print("\n전체 데이터 컬럼:")
print(all_combined.columns.tolist())

print("\n첫 5행:")
print(all_combined.head())

print("\n마지막 5행:")
print(all_combined.tail())
```

# Preprocessing

```python
all_combined['선거종류'].unique()
```

```python
mapping_dictionary = {
    '대통령': '대선',
    '광역단체장': '지선'
}

df_tidy = all_combined.replace({'선거종류': mapping_dictionary})

df_tidy
```

```python
df_tidy['지역'].unique()
```

```python
df_tidy = df_tidy[df_tidy['지역'] != '전국']
# df_tidy = df_tidy.query("지역 != '전국'")
```

```python
df_tidy.columns
```

```python
df_tidy = df_tidy.loc[:, '지역':'무소속_후보자수']
```

```python
df_tidy.info()
```

```python
df_tidy = df_tidy.astype({'선거년도': 'object'})
```

```python
df_tidy.info()
```

```python
df_tidy.to_csv("tidy2_combined.csv", index=False, encoding="utf-8-sig")
```
